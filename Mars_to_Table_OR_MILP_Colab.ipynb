# 🚀 Mars to Table × Investigación de Operaciones
## Optimización MILP de un sistema alimentario para una misión a Marte

**Caso de estudio de Operations Research (OR) / Supply Chain Analytics**

Este notebook transforma un problema de sostenibilidad alimentaria para una misión de larga duración en un **modelo de Programación Lineal Entera Mixta (MILP)** usando **PuLP + CBC**.

### Pregunta de decisión

> **¿Qué combinación de cultivos y fuentes alternativas de proteína debería operar una tripulación en Marte para reducir la dependencia de la Tierra, respetando simultáneamente límites de área, agua, biomasa, nutrición y diversidad?**

El modelo también explora el **trade-off entre autosuficiencia y diversidad** mediante distintos valores de `K`, el número mínimo de especies activas.

### Referencia oficial

NASA describe **Mars to Table** como un desafío para diseñar sistemas alimentarios capaces de sostener misiones prolongadas en la Luna y Marte, considerando nutrición, seguridad, usabilidad, integración con sistemas de soporte vital y operación del sistema.

### Modelo

```python
"""
Mars to Table — Modelo MILP corregido (PuLP), v3
================================================
Resuelve los problemas identificados en v2:

1. [ALTA] Proteinas alternativas AHORA aportan calorias (r_kcal_kg calculado
   de datos de literatura: FAO, USDA, estudios de insectos como alimento).

2. [ALTA] Seccion 6b duplicada renombrada -> 6d (agua neta con tasas globales).
   La seccion 6b original (tasas diferenciadas por candidato) se preserva.

3. [MEDIA] Validacion de datos de entrada: verifica positividad, maneja NaN
   con valores por defecto documentados, aborta con mensaje claro si hay
   datos criticos faltantes.

4. [MEDIA] Pesos de la funcion objetivo parametrizados (no mas magic numbers).

5. [MEDIA] Mejor documentacion de variables y unidades.

6. [BAJA] Graficos adicionales: evolucion de la mezcla para multiples K,
   sensibilidad del agua neta vs. tasa de reciclaje.

Lee 'candidatos_mars_to_table.xlsx' (13 cultivos fotosinteticos + 5 proteinas
alternativas, incluyendo Moringa oleifera como candidato de zona arida).
"""

import math
import sys
import pandas as pd
import pulp
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# EXCEL_PATH ya fue definido en la celda de carga de Colab.
```

### 0. PARAMETROS CONFIGURABLES (antes eran magic numbers)

```python
# ------------------------------------------------------------------
# 0. PARAMETROS CONFIGURABLES (antes eran magic numbers)
WEIGHT_EARTH_DEPENDENCY = 1000.0   # peso de la dependencia de Tierra en objetivo
WEIGHT_WATER = 0.001               # peso del agua en objetivo (tie-breaker)
BIG_M = 1e5                      # constante Big-M para activacion
MIN_AVG_KG_DIA = 0.05            # produccion minima promedio para que especie "cuente"
```

### 1. ESCENARIO DE MISION

```python
# ------------------------------------------------------------------
N_CREW = 4
T_WEEKS = 16
RDA_PROTEIN_G = 56
RDA_KCAL = 2500
A_MAX_M2 = 250
W_MAX_L_DIA = 2000
B_MAX_KG_DIA = 100
EARTH_CAP_START = 0.50
EARTH_CAP_END = 0.30
MIN_SPECIES_VALUES = [1, 3, 5, 7, 9, 11, 13, 15, 17]

d_protein = N_CREW * RDA_PROTEIN_G
d_kcal = N_CREW * RDA_KCAL


def earth_cap(t):
    """Fraccion maxima de deficit que puede venir de la Tierra en la semana t.
    Decae linealmente de EARTH_CAP_START a EARTH_CAP_END."""
    if T_WEEKS == 1:
        return EARTH_CAP_START
    frac = (t - 1) / (T_WEEKS - 1)
    return EARTH_CAP_START - (EARTH_CAP_START - EARTH_CAP_END) * frac
```

### 2. DATOS AUXILIARES: calorias de proteinas alternativas

```python
# ------------------------------------------------------------------
# Estos valores NO existen en el Excel original. Se calculan a partir de
# composicion nutricional tipica reportada en literatura:
#   - Grillo: ~65% proteina, ~20% grasa, ~5% carbohidratos -> ~4500 kcal/kg
#   - Tenebrio: ~50% proteina, ~30% grasa -> ~5500 kcal/kg
#   - Micoproteina (Quorn): ~45% proteina, bajo grasa -> ~1100 kcal/kg
#   - Hongo ostra (fresco): ~3% proteina, ~1% grasa, ~5% carbs -> ~330 kcal/kg
#   - Carne cultivada: ~20% proteina, ~20% grasa -> ~2500 kcal/kg
# Fuentes: FAO Nutritional Database, USDA FoodData Central, Oonincx et al. 2015
# (J. Insects Food Feed), Finke 2002 (complete nutrient composition of insects).
ALT_KCAL_PER_KG = {
    "Grillo (Acheta domesticus)": 4500,
    "Tenebrio / gusano de la harina (Tenebrio molitor)": 5500,
    "Micoproteina / Fusarium venenatum (tipo Quorn)": 1100,
    "Hongo ostra / Oyster mushroom (Pleurotus ostreatus)": 330,
    "Carne cultivada / Cultivated meat (celulas animales in vitro)": 2500,
}

# Valor por defecto para agua de hongo ostra (NaN en Excel original).
# Estimado por analogia con micoproteina (ambos son hongos, cultivo en
# sustrato/solucion nutritiva). La literatura de Pleurotus reporta
# 0.42-1.26 m3/kg en base seca; como se consume fresco (~90% agua),
# el valor por kg fresco es menor. Usamos 1500 L/kg como estimacion
# conservadora hasta obtener dato experimental.
DEFAULT_WATER_OYSTER_MUSHROOM = 1500.0
```

### 3. CARGA DE DATOS CON VALIDACION

```python
# ------------------------------------------------------------------
def load_and_validate_data(excel_path):
    """Carga datos del Excel y valida consistencia. Aborta con mensaje
    claro si hay datos criticos faltantes o invalidos."""

    df1 = pd.read_excel(excel_path, sheet_name="Cultivos_Fotosinteticos", header=3).dropna(subset=["Candidato"])
    df2 = pd.read_excel(excel_path, sheet_name="Proteinas_Alternativas", header=3).dropna(subset=["Candidato"])

    crops, alt = {}, {}
    errors = []
    warnings = []

    # --- Validar cultivos fotosinteticos ---
    required_cols_crops = {
        "r_proteina_g_m2_d": float,
        "r_calorico_kcal_m2_d": float,
        "a_i_m2_por_kg_comestible_dia": float,
        "w_i": "Agua_L_m2_d",
        "edible_g_m2d": "Biomasa_comestible_gDM_m2_d",
        "ciclo_dias": "Ciclo_dias",
    }

    for _, row in df1.iterrows():
        name = row["Candidato"]

        # Verificar campos numericos criticos
        r_prot = float(row["r_proteina_g_m2_d"])
        r_kcal = float(row["r_calorico_kcal_m2_d"])
        a_i = float(row["a_i_m2_por_kg_comestible_dia"])
        w_i = float(row["Agua_L_m2_d"])
        edible = float(row["Biomasa_comestible_gDM_m2_d"])
        ciclo = float(row["Ciclo_dias"])

        for label, val in [("r_proteina", r_prot), ("r_calorico", r_kcal),
                           ("a_i", a_i), ("w_i", w_i), ("edible", edible),
                           ("ciclo_dias", ciclo)]:
            if math.isnan(val) or val is None:
                errors.append(f"[{name}] {label} = NaN/None (dato critico faltante)")
            elif val < 0:
                errors.append(f"[{name}] {label} = {val} (negativo, debe ser >= 0)")
            elif val == 0 and label in ("r_proteina", "r_calorico", "edible"):
                warnings.append(f"[{name}] {label} = 0 (cultivo no aporta este nutriente)")

        if ciclo <= 0:
            errors.append(f"[{name}] ciclo_dias = {ciclo} (debe ser > 0)")

        # Verificar consistencia: concentracion de proteina calculada vs. % DM
        if edible > 0:
            prot_conc_calc = r_prot / (edible / 1000)  # g proteina / kg biomasa
            prot_pct_dm = row.get("Proteina_pct_DM", float("nan"))
            if not math.isnan(prot_pct_dm):
                prot_conc_expected = float(prot_pct_dm) * 10  # g/kg
                if abs(prot_conc_calc - prot_conc_expected) > prot_conc_expected * 0.5:
                    warnings.append(
                        f"[{name}] Inconsistencia proteina: calculada={prot_conc_calc:.1f} g/kg "
                        f"vs. Proteina_pct_DM*10={prot_conc_expected:.1f} g/kg"
                    )

        crops[name] = {
            "tipo": "fotosintetico",
            "r_prot_m2d": r_prot,
            "r_kcal_m2d": r_kcal,
            "a_i": a_i,
            "w_i": w_i,
            "edible_g_m2d": edible,
            "ciclo_dias": ciclo,
        }

    # --- Validar proteinas alternativas ---
    for _, row in df2.iterrows():
        name = row["Candidato"]

        r_prot_kg = float(row["r_proteina_g_por_kg_producto"])
        agua = row["Agua_L_por_kg_producto"]
        ciclo = float(row["Ciclo_dias"])

        # Manejar NaN en agua (hongo ostra)
        if pd.isna(agua):
            agua = DEFAULT_WATER_OYSTER_MUSHROOM
            warnings.append(
                f"[{name}] Agua_L_por_kg_producto = NaN en Excel. "
                f"Usando valor por defecto {DEFAULT_WATER_OYSTER_MUSHROOM} L/kg "
                f"(estimado por analogia con micoproteina, validar experimentalmente)."
            )
        else:
            agua = float(agua)

        # Verificar campos criticos
        for label, val in [("r_proteina_g_por_kg", r_prot_kg), ("ciclo_dias", ciclo)]:
            if math.isnan(val) or val is None:
                errors.append(f"[{name}] {label} = NaN/None (dato critico faltante)")
            elif val < 0:
                errors.append(f"[{name}] {label} = {val} (negativo)")

        if ciclo <= 0:
            errors.append(f"[{name}] ciclo_dias = {ciclo} (debe ser > 0)")

        # Verificar que tenemos calorias para este candidato
        if name not in ALT_KCAL_PER_KG:
            errors.append(
                f"[{name}] No existe entrada en ALT_KCAL_PER_KG. "
                f"Agregar calorias estimadas (kcal/kg producto) al diccionario."
            )
            r_kcal_kg = 0.0
        else:
            r_kcal_kg = ALT_KCAL_PER_KG[name]

        # Verificar consistencia: r_proteina_g_por_kg vs. Proteina_pct_producto
        prot_pct = row.get("Proteina_pct_producto", float("nan"))
        if not math.isnan(prot_pct):
            expected_prot = float(prot_pct) * 10
            if abs(r_prot_kg - expected_prot) > 1.0:
                warnings.append(
                    f"[{name}] Inconsistencia: r_proteina_g_por_kg={r_prot_kg} "
                    f"vs. Proteina_pct_producto*10={expected_prot}"
                )

        alt[name] = {
            "tipo": "alternativo",
            "r_prot_kg": r_prot_kg,
            "r_kcal_kg": r_kcal_kg,  # <-- NUEVO: calorias por kg de producto
            "w_i": agua,
            "ciclo_dias": ciclo,
        }

    # Reportar y abortar si hay errores criticos
    if warnings:
        print("\n=== ADVERTENCIAS (no bloqueantes) ===")
        for w in warnings:
            print(f"  ⚠️  {w}")

    if errors:
        print("\n=== ERRORES CRITICOS (bloqueantes) ===")
        for e in errors:
            print(f"  ❌  {e}")
        print("\n>>> Abortando: corrige los errores en el Excel o el codigo antes de continuar.")
        sys.exit(1)

    print(f"\n✅ Validacion exitosa: {len(crops)} cultivos fotosinteticos y {len(alt)} proteinas alternativas.")
    return crops, alt


crops, alt = load_and_validate_data(EXCEL_PATH)
all_ids = list(crops.keys()) + list(alt.keys())
weeks = list(range(1, T_WEEKS + 1))
print(f"Candidatos totales: {len(all_ids)}.")


def first_week(cycle_days):
    """Primera semana en la que el cultivo puede producir (tras madurar)."""
    return math.ceil(cycle_days / 7)
```

### 4. FUNCION QUE ARMA Y RESUELVE EL MODELO PARA UN K DADO

```python
# ------------------------------------------------------------------
def solve_model(min_species):
    """Resuelve el modelo MILP para un valor dado de K (especies minimas).

    Variables:
      x[(i,t)]  : produccion de candidato i en semana t (kg/dia de biomasa
                  comestible para fotosinteticos; kg/dia de producto para
                  alternativos)
      y[i]      : 1 si el candidato i esta activo, 0 en caso contrario
      delta_prot[t] : deficit de proteina cubierto por Tierra en semana t (g/dia)
      delta_kcal[t] : deficit de calorias cubierto por Tierra en semana t (kcal/dia)

    La funcion objetivo minimiza la dependencia de la Tierra (dominante) y
    el consumo de agua (secundario, tie-breaker).
    """
    prob = pulp.LpProblem(f"Mars_to_Table_K{min_species}", pulp.LpMinimize)

    x = {(i, t): pulp.LpVariable(f"x_{i}_{t}_{min_species}", lowBound=0)
         for i in all_ids for t in weeks}
    y = {i: pulp.LpVariable(f"y_{i}_{min_species}", cat="Binary")
         for i in all_ids}
    delta_prot = {t: pulp.LpVariable(f"dp_{t}_{min_species}", lowBound=0)
                  for t in weeks}
    delta_kcal = {t: pulp.LpVariable(f"dk_{t}_{min_species}", lowBound=0)
                  for t in weeks}

    # --- Terminos de agua ---
    water_terms = (
        [crops[i]["w_i"] * crops[i]["a_i"] * x[(i, t)]
         for i in crops for t in weeks]
        + [alt[i]["w_i"] * x[(i, t)]
           for i in alt for t in weeks]
    )

    # --- Dependencia de la Tierra ---
    earth_dependency = (
        pulp.lpSum(delta_prot[t] for t in weeks) / d_protein
        + pulp.lpSum(delta_kcal[t] for t in weeks) / d_kcal
    )

    # --- Funcion objetivo ---
    prob += (WEIGHT_EARTH_DEPENDENCY * earth_dependency
             + WEIGHT_WATER * pulp.lpSum(water_terms)), "Objetivo"

    # --- Restricciones semanales ---
    for t in weeks:
        # Suministro de proteina (g/dia)
        prot_supply = (
            pulp.lpSum(
                (crops[i]["r_prot_m2d"] / (crops[i]["edible_g_m2d"] / 1000))
                * x[(i, t)]
                for i in crops
            )
            + pulp.lpSum(alt[i]["r_prot_kg"] * x[(i, t)] for i in alt)
        )
        prob += prot_supply + delta_prot[t] >= d_protein
        prob += delta_prot[t] <= earth_cap(t) * d_protein

        # Suministro de calorias (kcal/dia)  <-- CORREGIDO: ahora incluye alternativos
        kcal_supply = (
            pulp.lpSum(
                (crops[i]["r_kcal_m2d"] / (crops[i]["edible_g_m2d"] / 1000))
                * x[(i, t)]
                for i in crops
            )
            + pulp.lpSum(alt[i]["r_kcal_kg"] * x[(i, t)] for i in alt)  # NUEVO
        )
        prob += kcal_supply + delta_kcal[t] >= d_kcal
        prob += delta_kcal[t] <= earth_cap(t) * d_kcal

        # Recursos
        prob += pulp.lpSum(crops[i]["a_i"] * x[(i, t)] for i in crops) <= A_MAX_M2
        prob += pulp.lpSum(x[(i, t)] for i in alt) <= B_MAX_KG_DIA
        prob += (
            pulp.lpSum(crops[i]["w_i"] * crops[i]["a_i"] * x[(i, t)]
                       for i in crops)
            + pulp.lpSum(alt[i]["w_i"] * x[(i, t)] for i in alt)
        ) <= W_MAX_L_DIA

        # Ciclo de cultivo y activacion
        for i in all_ids:
            src = crops[i] if i in crops else alt[i]
            fw = first_week(src["ciclo_dias"])
            if t < fw:
                prob += x[(i, t)] == 0
            prob += x[(i, t)] <= BIG_M * y[i]

    # --- Restriccion de diversidad: produccion minima por especie activa ---
    for i in all_ids:
        src = crops[i] if i in crops else alt[i]
        fw = first_week(src["ciclo_dias"])
        active_weeks = [t for t in weeks if t >= fw]
        if active_weeks:
            prob += (pulp.lpSum(x[(i, t)] for t in active_weeks)
                     >= MIN_AVG_KG_DIA * len(active_weeks) * y[i])

    # --- Restriccion de diversidad: minimo de especies activas ---
    prob += pulp.lpSum(y[i] for i in all_ids) >= min_species, "Diversidad_minima"

    # --- Resolver ---
    status = prob.solve(pulp.PULP_CBC_CMD(msg=0))

    result = {
        "K": min_species,
        "status": pulp.LpStatus[status],
        "model": prob,
        "vars": {"x": x, "y": y, "delta_prot": delta_prot, "delta_kcal": delta_kcal},
    }

    if pulp.LpStatus[status] == "Optimal":
        result["earth_dependency"] = pulp.value(earth_dependency)
        result["agua_total"] = sum(pulp.value(term) for term in water_terms)
        active = [i for i in all_ids if (y[i].value() or 0) > 0.5]
        result["especies_activas"] = active
        result["n_especies"] = len(active)
        result["weekly"] = {(i, t): (x[(i, t)].value() or 0)
                            for i in all_ids for t in weeks}
        # NUEVO: metricas adicionales
        result["area_usada_prom"] = sum(
            pulp.lpSum(crops[i]["a_i"] * x[(i, t)] for i in crops).value()
            for t in weeks
        ) / T_WEEKS
        result["biomasa_alt_prom"] = sum(
            pulp.lpSum(x[(i, t)] for i in alt).value()
            for t in weeks
        ) / T_WEEKS
    else:
        result["earth_dependency"] = None
        result["agua_total"] = None
        result["especies_activas"] = []
        result["n_especies"] = 0
        result["weekly"] = {}
        result["area_usada_prom"] = None
        result["biomasa_alt_prom"] = None

    return result
```

### 5. CORRER EL MODELO PARA CADA VALOR DE K

```python
# ------------------------------------------------------------------
resultados = [solve_model(k) for k in MIN_SPECIES_VALUES]

print("\n=== Comparacion: costo de forzar diversidad de especies ===")
comp_rows = []
for r in resultados:
    comp_rows.append({
        "K_especies_minimas": r["K"],
        "Estado": r["status"],
        "Especies_activas": r["n_especies"],
        "Dependencia_Tierra": round(r["earth_dependency"], 4)
        if r["earth_dependency"] is not None else None,
        "Agua_total_prom_L_dia": round(r["agua_total"] / T_WEEKS, 1)
        if r["agua_total"] is not None else None,
        "Area_usada_prom_m2": round(r["area_usada_prom"], 1)
        if r["area_usada_prom"] is not None else None,
        "Biomasa_alt_prom_kg_dia": round(r["biomasa_alt_prom"], 2)
        if r["biomasa_alt_prom"] is not None else None,
        "Lista_especies": ", ".join(r["especies_activas"])
        if r["especies_activas"] else "-",
    })
comp_df = pd.DataFrame(comp_rows)
print(comp_df.drop(columns=["Lista_especies"]).to_string(index=False))
print()
for r in resultados:
    print(f"K={r['K']:>2} ({r['status']}): "
          f"{', '.join(r['especies_activas']) if r['especies_activas'] else '(infactible)'}")
```

### 6a. AGUA NETA DE REPOSICION CON TASAS DE RECICLAJE DIFERENCIADAS POR CANDIDATO

```python
# ------------------------------------------------------------------
# Principio fisico clave para un habitat SELLADO: el agua que se evapora
# de cultivos fotosinteticos NO se pierde a la atmosfera exterior -- queda
# como humedad dentro del habitat y el HVAC la recupera. Las proteinas
# alternativas tienen tasas mas bajas porque parte del agua queda embebida
# en residuos solidos (sustrato de hongo, frass de insectos, medio celular).
#
# IMPORTANTE: estas tasas son ESTIMACIONES RAZONADAS a partir de principios
# de ECLSS/raceways -- no existe un estudio que mida "tasa de reciclaje de
# agua por candidato de sistema alimentario espacial". Tratarlas como
# supuestos de ingenieria a validar, no como dato duro.
RECYCLE_RATE_BY_CANDIDATE = {
    # --- Cultivos fotosinteticos hidroponicos ---
    "Trigo (Triticum aestivum)": 0.96,
    "Soya (Glycine max)": 0.96,
    "Papa (Solanum tuberosum)": 0.96,
    "Camote (Ipomoea batatas)": 0.96,
    "Lechuga (Lactuca sativa)": 0.96,
    "Tomate (Solanum lycopersicum)": 0.96,
    "Arroz (Oryza sativa)": 0.95,
    "Mani / Peanut (Arachis hypogaea)": 0.95,
    "Quinua (Chenopodium quinoa)": 0.95,
    "Moringa (Moringa oleifera)": 0.95,
    # --- Microalgas/macrofitas acuaticas ---
    "Espirulina (Arthrospira platensis)": 0.94,
    "Chlorella (Chlorella vulgaris)": 0.94,
    "Lenteja de agua / Duckweed (Lemna minor)": 0.94,
    # --- Proteinas alternativas ---
    "Grillo (Acheta domesticus)": 0.75,
    "Tenebrio / gusano de la harina (Tenebrio molitor)": 0.75,
    "Micoproteina / Fusarium venenatum (tipo Quorn)": 0.80,
    "Hongo ostra / Oyster mushroom (Pleurotus ostreatus)": 0.65,
    "Carne cultivada / Cultivated meat (celulas animales in vitro)": 0.70,
}
DEFAULT_RECYCLE_RATE = 0.90

print("\n=== 6a. Agua NETA de reposicion con tasa de reciclaje diferenciada por candidato (K=7) ===")
k7_result = next(r for r in resultados if r["K"] == 7)
per_candidate_rows = []
total_bruta = 0.0
total_neta = 0.0
for i in all_ids:
    bruta_i = 0.0
    for t in weeks:
        v = k7_result["weekly"].get((i, t), 0)
        if i in crops:
            bruta_i += crops[i]["w_i"] * crops[i]["a_i"] * v
        else:
            bruta_i += alt[i]["w_i"] * v
    if bruta_i <= 1e-9:
        continue
    rate = RECYCLE_RATE_BY_CANDIDATE.get(i, DEFAULT_RECYCLE_RATE)
    neta_i = bruta_i * (1 - rate)
    total_bruta += bruta_i
    total_neta += neta_i
    per_candidate_rows.append({
        "Candidato": i,
        "Agua_bruta_16sem_L": round(bruta_i * 7, 1),
        "Tasa_reciclaje_pct": round(rate * 100, 1),
        "Agua_neta_16sem_L": round(neta_i * 7, 1),
    })

per_df = pd.DataFrame(per_candidate_rows).sort_values("Agua_bruta_16sem_L", ascending=False)
print(per_df.to_string(index=False))
tasa_efectiva = 1 - (total_neta / total_bruta) if total_bruta else 0
print(f"\nTotal agua bruta (16 sem): {total_bruta*7:,.0f} L (~{total_bruta*7/1000:,.2f} m3)")
print(f"Total agua NETA de reposicion (16 sem, tasas diferenciadas): {total_neta*7:,.0f} L (~{total_neta*7/1000:,.2f} m3)")
print(f"Tasa de reciclaje EFECTIVA del sistema completo (ponderada): {tasa_efectiva*100:.1f}%")
```

### 6b. BALANCE HIDRICO COMPLETO: transpiracion vs. agua embebida en cosecha

```python
# ------------------------------------------------------------------
# w_i es el agua TOTAL entregada al sistema hidroponico (riego + solucion
# nutritiva + acido de pH) segun metodologia NASA (Wheeler et al. 1999).
# Se separa en:
#   1) Agua TRANSPIRADA: se condensa en HVAC y se recicla -> tasa RECYCLE_RATE.
#   2) Agua EMBEBIDA en biomasa COSECHADA: sale con el alimento, se recupera
#      via ECLSS humano (ISS 2023: 98% recuperacion).
# Los valores de Biomasa_comestible_gDM_m2_d estan en base SECA (NASA, 70C/72h).
DM_FRACTION = {
    "Trigo (Triticum aestivum)": 0.88,
    "Soya (Glycine max)": 0.90,
    "Papa (Solanum tuberosum)": 0.20,
    "Camote (Ipomoea batatas)": 0.23,
    "Lechuga (Lactuca sativa)": 0.05,
    "Tomate (Solanum lycopersicum)": 0.06,
    "Arroz (Oryza sativa)": 0.88,
    "Mani / Peanut (Arachis hypogaea)": 0.92,
    "Quinua (Chenopodium quinoa)": 0.88,
    "Moringa (Moringa oleifera)": 0.90,
    "Espirulina (Arthrospira platensis)": 0.92,
    "Chlorella (Chlorella vulgaris)": 0.92,
    "Lenteja de agua / Duckweed (Lemna minor)": 0.10,
}
ECLSS_HUMANO_RECOVERY = 0.98

print("\n=== 6b. Balance hidrico completo K=7: transpiracion vs. embebida ===")
balance_rows = []
total_transp_bruta = total_transp_neta = 0.0
total_embed_bruta = total_embed_neta = 0.0
for i in all_ids:
    if i not in crops:
        continue
    dry_kg_16sem = sum(k7_result["weekly"].get((i, t), 0) for t in weeks) * 7
    if dry_kg_16sem <= 1e-9:
        continue
    dm = DM_FRACTION.get(i, 0.90)
    fresh_kg_16sem = dry_kg_16sem / dm
    embebida_L = fresh_kg_16sem - dry_kg_16sem
    bruta_total_i = next(
        row["Agua_bruta_16sem_L"] for row in per_candidate_rows if row["Candidato"] == i
    )
    transp_L = max(bruta_total_i - embebida_L, 0)

    rate_transp = RECYCLE_RATE_BY_CANDIDATE.get(i, DEFAULT_RECYCLE_RATE)
    transp_neta = transp_L * (1 - rate_transp)
    embebida_neta = embebida_L * (1 - ECLSS_HUMANO_RECOVERY)

    total_transp_bruta += transp_L
    total_transp_neta += transp_neta
    total_embed_bruta += embebida_L
    total_embed_neta += embebida_neta

    balance_rows.append({
        "Candidato": i,
        "Agua_transpirada_16sem_L": round(transp_L, 1),
        "Agua_embebida_en_cosecha_16sem_L": round(embebida_L, 1),
        "pct_embebida_del_total": round(100 * embebida_L / bruta_total_i, 2)
        if bruta_total_i else 0,
        "Neta_transpiracion_L": round(transp_neta, 2),
        "Neta_embebida_L": round(embebida_neta, 2),
    })

balance_df = pd.DataFrame(balance_rows)
print(balance_df.to_string(index=False))

total_bruta_16 = total_transp_bruta + total_embed_bruta
total_neta_16 = total_transp_neta + total_embed_neta
print(f"\nAgua transpirada (recirculada en habitat): "
      f"{total_transp_bruta:,.0f} L bruta -> {total_transp_neta:,.1f} L neta")
print(f"Agua embebida en cosecha (recirculada via ECLSS tripulacion): "
      f"{total_embed_bruta:,.1f} L bruta -> {total_embed_neta:,.2f} L neta")
print(f"TOTAL agua neta de reposicion (ambas rutas): {total_neta_16:,.1f} L "
      f"(~{total_neta_16/1000:,.3f} m3) en 16 semanas")
```

### 6c. AGUA NETA DE REPOSICION: sensibilidad a tasas de reciclaje GLOBALES

```python
# ------------------------------------------------------------------
# Enfoque simplificado: una tasa global de reciclaje para todo el sistema.
# Util para analisis de sensibilidad y comparacion con literatura de BLSS.
WATER_RECYCLE_RATES = [0.90, 0.93, 0.95, 0.98]

print("\n=== 6c. Agua NETA de reposicion segun tasa de recuperacion GLOBAL del sistema ===")
net_rows = []
for r in resultados:
    if r["status"] != "Optimal":
        continue
    # r["agua_total"] = suma sobre T_WEEKS semanas del agua diaria (L/dia * 1 semana)
    # = promedio diario * T_WEEKS (porque cada semana suma 7 * promedio_diario_semana)
    # Pero en realidad es: para cada semana t, sumamos agua_dia_t * 1 (la unidad es semana)
    # Entonces agua_total / T_WEEKS = promedio diario sobre toda la mision
    agua_bruta_dia = r["agua_total"] / T_WEEKS
    agua_bruta_total_16sem = agua_bruta_dia * T_WEEKS * 7  # 7 dias/semana
    row = {
        "K": r["K"],
        "Agua_bruta_L_dia": round(agua_bruta_dia, 1),
        "Agua_bruta_total_16sem_L": round(agua_bruta_total_16sem, 0),
    }
    for rate in WATER_RECYCLE_RATES:
        row[f"Neta_{int(rate*100)}pct_L_dia"] = round(agua_bruta_dia * (1 - rate), 2)
        row[f"Neta_{int(rate*100)}pct_total_16sem_L"] = round(
            agua_bruta_total_16sem * (1 - rate), 1
        )
    net_rows.append(row)

net_df = pd.DataFrame(net_rows)
cols_display = ["K", "Agua_bruta_L_dia"] + [f"Neta_{int(r*100)}pct_L_dia"
                                            for r in WATER_RECYCLE_RATES]
print(net_df[cols_display].to_string(index=False))

k7 = next((r for r in net_rows if r["K"] == 7), None)
if k7:
    print(f"\n--- Detalle K=7 (escenario recomendado), 16 semanas ---")
    print(f"Agua bruta (throughput): {k7['Agua_bruta_total_16sem_L']:,.0f} L "
          f"(~{k7['Agua_bruta_total_16sem_L']/1000:,.1f} m3)")
    for rate in WATER_RECYCLE_RATES:
        total_neta = k7[f"Neta_{int(rate*100)}pct_total_16sem_L"]
        dia_neta = k7[f"Neta_{int(rate*100)}pct_L_dia"]
        print(f"  Con {int(rate*100)}% reciclaje -> reposicion neta: "
              f"{dia_neta:,.2f} L/dia | {total_neta:,.0f} L total "
              f"({total_neta/1000:,.2f} m3)")
```

### 6d. COMPARACION: tasas diferenciadas vs. tasa global uniforme

```python
# ------------------------------------------------------------------
# NUEVO: compara el resultado del balance hidrico completo (6b) con el
# supuesto simplificado de tasa global uniforme.
print("\n=== 6d. Comparacion: balance hidrico completo vs. tasa global uniforme ===")
if k7:
    # Agua neta del balance completo (seccion 6b)
    neta_balance_completo = total_neta_16
    # Agua neta con tasa global uniforme de 95% (supuesto original v1)
    agua_bruta_total = k7["Agua_bruta_total_16sem_L"]
    neta_global_95 = agua_bruta_total * (1 - 0.95)
    neta_global_90 = agua_bruta_total * (1 - 0.90)
    neta_global_98 = agua_bruta_total * (1 - 0.98)

    print(f"Agua bruta total (K=7): {agua_bruta_total:,.0f} L")
    print(f"  -> Balance hidrico COMPLETO (6b): {neta_balance_completo:,.1f} L neta")
    print(f"  -> Supuesto global 95% uniforme:  {neta_global_95:,.0f} L neta")
    print(f"  -> Supuesto global 90% uniforme:  {neta_global_90:,.0f} L neta")
    print(f"  -> Supuesto global 98% uniforme:  {neta_global_98:,.0f} L neta")

    tasa_global_equivalente = 1 - (neta_balance_completo / agua_bruta_total)
    print(f"\nTasa de reciclaje GLOBAL equivalente al balance completo: "
          f"{tasa_global_equivalente*100:.1f}%")
    print(f"(La mezcla optima K=7 tiene una tasa efectiva de {tasa_global_equivalente*100:.1f}%, "
          f"que es {'mayor' if tasa_global_equivalente > 0.95 else 'menor'} que el supuesto "
          f"simplificado de 95% uniforme)")
```

# 🧠 Lectura OR del modelo

## Función objetivo

El modelo minimiza principalmente la dependencia de recursos provenientes de la Tierra y utiliza el agua como criterio secundario:

\[
\min \quad
w_E \cdot 	ext{EarthDependency}
+
w_W \cdot 	ext{WaterUse}
\]

donde los pesos son parámetros explícitos y no constantes ocultas.

## Decisiones

\[
x_{i,t} \ge 0
\]

representa la producción diaria del candidato \(i\) en la semana \(t\).

\[
y_i \in \{0,1\}
\]

indica si el candidato forma parte del sistema.

## Restricciones principales

### Nutrición

\[
ProteinSupply_t + \Delta Protein_t \ge DemandProtein
\]

\[
CaloriesSupply_t + \Delta Calories_t \ge DemandCalories
\]

### Dependencia de la Tierra

El déficit nutricional cubierto desde Tierra queda limitado por una capacidad que disminuye durante la misión.

### Recursos

- Área máxima
- Agua máxima diaria
- Biomasa máxima de fuentes alternativas de proteína

### Maduración

Un candidato no puede producir antes de completar su ciclo de crecimiento.

### Diversidad

\[
\sum_i y_i \ge K
\]

Esto convierte la diversidad en una decisión estratégica y permite estudiar su **costo de oportunidad**.

---

## 🔬 Insight de OR

La pregunta interesante no es simplemente:

> “¿Cuál cultivo produce más?”

sino:

> **“¿Qué portafolio de producción satisface simultáneamente múltiples restricciones y qué precio operativo tiene exigir resiliencia/diversidad?”**

Ese cambio de perspectiva es precisamente lo que convierte un problema espacial en un **caso de decisión cuantitativa**.


# 📊 Cómo interpretar los resultados

Al ejecutar el notebook, presta especial atención a:

### 1. `Dependencia_Tierra`
Cuanto menor, mayor autosuficiencia del sistema.

### 2. `Agua_total_prom_L_dia`
Permite comparar la eficiencia hídrica de distintos niveles de diversidad.

### 3. `Area_usada_prom_m2`
Mide utilización de la capacidad de cultivo.

### 4. `Biomasa_alt_prom_kg_dia`
Muestra cuánto depende la solución de fuentes alternativas de proteína.

### 5. `K`
Permite observar el trade-off entre **eficiencia** y **resiliencia/diversidad**.

Una solución con más especies puede ser estratégicamente preferible aunque tenga un costo marginal de recursos. En un sistema crítico, esa diferencia puede ser más importante que minimizar un único KPI.


# ⚠️ Supuestos que deben tratarse como hipótesis

Para presentar este caso profesionalmente, es importante distinguir entre **datos** y **supuestos de modelado**.

El propio código identifica como estimaciones, entre otros:

- calorías de proteínas alternativas;
- tasa de reciclaje de agua por candidato;
- estimación de agua para hongo ostra cuando falta el dato;
- ciertos parámetros nutricionales.

Por tanto, los resultados son **resultados del modelo bajo los supuestos introducidos**, no predicciones experimentales.

Esto es una fortaleza para un proyecto de OR: un buen optimizador no solo entrega una solución; también identifica **qué supuestos pueden cambiar la decisión**.


# 🌎 Extensión recomendada

Este modelo puede evolucionar hacia un proyecto de investigación más potente:

1. **Robust Optimization:** incertidumbre en rendimientos de cultivos.

4. **Scenario Analysis:** fallos de cultivos o restricciones de energía.
5. **Monte Carlo:** sensibilidad probabilística.
6. **Dynamic Optimization:** decisiones de producción a lo largo de la misión.

8. **Reinforcement Learning:** política adaptativa ante cambios del sistema.
9. **Supply Chain Design:** incorporar inventarios, resupply y lead times Tierra–Marte.

Esto convierte el notebook en el punto de partida de un **framework de Decision Science para sistemas autónomos y supply chains extremas**.
